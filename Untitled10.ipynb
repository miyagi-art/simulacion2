```python

import numpy as np
import matplotlib.pyplot as plt

# Definición de la función
def f(x1, x2, C=1.0):
    exponent = -(x1**2 * x2**2 + x1**2 + x2**2 - 8*x1 - 8*x2) / 2
    return C * np.exp(exponent)

# Crear la malla de puntos (x1, x2)
x1 = np.linspace(-3, 5, 300)
x2 = np.linspace(-3, 5, 300)
X1, X2 = np.meshgrid(x1, x2)
Z = f(X1, X2)

# Crear figura con 2 subgráficas
fig = plt.figure(figsize=(14, 6))

# 1. Gráfica en 3D (Superficie)
ax1 = fig.add_subplot(1, 2, 1, projection='3d')
surf = ax1.plot_surface(X1, X2, Z, cmap='viridis', edgecolor='none', alpha=0.9)
ax1.set_title(r'Gráfica 3D de $f(x_1, x_2)$')
ax1.set_xlabel(r'$x_1$')
ax1.set_ylabel(r'$x_2$')
ax1.set_zlabel(r'$f(x_1, x_2)$')
fig.colorbar(surf, ax=ax1, shrink=0.5, aspect=10)

# 2. Mapa de contorno (2D)
ax2 = fig.add_subplot(1, 2, 2)
contour = ax2.contourf(X1, X2, Z, levels=30, cmap='viridis')
ax2.set_title(r'Mapa de Contornos de $f(x_1, x_2)$')
ax2.set_xlabel(r'$x_1$')
ax2.set_ylabel(r'$x_2$')
ax2.grid(True, linestyle='--', alpha=0.5)
fig.colorbar(contour, ax=ax2)

plt.tight_layout()
plt.show()
```
