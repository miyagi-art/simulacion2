```python

import numpy as np
import matplotlib.pyplot as plt

# 1. Definición de la función objetivo f(x1, x2)
def f(x1, x2):
    c = 1 / 20216.335877
    exponent = -(x1**2 * x2**2 + x1**2 + x2**2 - 8*x1 - 8*x2) / 2
    return c * np.exp(exponent)

# Función de probabilidad de aceptación alpha(X, Y)
def calcular_alpha(x_t, y):
    fx_t = f(x_t[0], x_t[1])
    if fx_t == 0:
        return 1.0
    return min(1.0, f(y[0], y[1]) / fx_t)

# 2. Algoritmo de Metropolis-Hastings (Caminata Aleatoria)
def caminata_aleatoria(N, x1_inicial, x2_inicial):
    muestras = np.zeros((N, 2))
    muestras[0] = np.array([x1_inicial, x2_inicial])

    for t in range(0, N - 1):
        x_t = muestras[t]

        # Propuesta Z ~ N(0,1)
        z = np.random.normal(0, 1, size=2)
        y = x_t + z

        # Calcular probabilidad de aceptación
        alpha = calcular_alpha(x_t, y)

        # Criterio de aceptación
        u = np.random.uniform(0, 1)
        if u < alpha:
            muestras[t + 1] = y
        else:
            muestras[t + 1] = x_t

    return muestras

# --- EJECUCIÓN Y SIMULACIÓN ---
N_muestras = 30000
punto_inicial = [0.0, 0.0]

muestras = caminata_aleatoria(N_muestras, punto_inicial[0], punto_inicial[1])

# --- GRAFICACIÓN ---

# Crear malla de puntos para graficar la función teórica f(x1, x2)
x1_grid = np.linspace(-1, 5, 300)
x2_grid = np.linspace(-1, 5, 300)
X1, X2 = np.meshgrid(x1_grid, x2_grid)
Z = f(X1, X2)

# FIGURA 1: Mapa de Contorno en 2D con los puntos simulados sobrepuestos
plt.figure(figsize=(9, 7))
contour = plt.contourf(X1, X2, Z, levels=30, cmap='viridis', alpha=0.8)
plt.colorbar(contour, label='Densidad de probabilidad $f(x_1, x_2)$')

# Se grafica un subconjunto de puntos (1 de cada 3) para evitar saturar la imagen
plt.scatter(muestras[::3, 0], muestras[::3, 1], c='red', s=2, alpha=0.3, label='Muestras de la caminata')

plt.title('Simulación Metropolis-Hastings sobre la Densidad Teórica')
plt.xlabel('$x_1$')
plt.ylabel('$x_2$')
plt.xlim(-1, 5)
plt.ylim(-1, 5)
plt.legend(loc='upper right')
plt.grid(True, linestyle='--', alpha=0.3)
plt.show()

# FIGURA 2: Superficie en 3D de la función f(x1, x2)
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')
surf = ax.plot_surface(X1, X2, Z, cmap='viridis', edgecolor='none', alpha=0.9)
fig.colorbar(surf, ax=ax, shrink=0.5, aspect=5, label='f(x1, x2)')

ax.set_title('Superficie 3D de la Función Objetivo $f(x_1, x_2)$')
ax.set_xlabel('$x_1$')
ax.set_ylabel('$x_2$')
ax.set_zlabel('Densidad')
plt.show()
```
