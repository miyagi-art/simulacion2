```python

from scipy.stats import norm

# Parámetros del problema[span_7](start_span)[span_7](end_span)
mu_param = 0.6  # Parámetro mu[span_8](start_span)[span_8](end_span)
sigma_cuadrada = 0.25  # Varianza sigma^2[span_9](start_span)[span_9](end_span)
sigma = sigma_cuadrada**0.5  # Desviación estándar (0.5)
t = 4  # Tiempo t[span_10](start_span)[span_10](end_span)

# Límites del intervalo de probabilidad[span_11](start_span)[span_11](end_span)
a = 1
b = 3

# Cálculo de la media y la desviación estándar para X_t al tiempo t
media = mu_param * t  # E[X_t] = 0.6 * 4 = 2.4
varianza = sigma_cuadrada * t  # Var(X_t) = 0.25 * 4 = 1.0
desviacion_estandar = varianza**0.5  # sqrt(1.0) = 1.0

# Cálculo de la probabilidad P(a <= X_t <= b) usando la distribución normal
probabilidad = norm.cdf(b, loc=media, scale=desviacion_estandar) - norm.cdf(
    a, loc=media, scale=desviacion_estandar
)

print(f"Media al tiempo t={t}: {media}")
print(f"Varianza al tiempo t={t}: {varianza}")
print(
    f"La probabilidad P({a} <= X_{t} <= {b}) es: {probabilidad:.6f}"
)  #[span_12](start_span)[span_12](end_span)
```

```python

import numpy as np

# Parámetros del problema
mu = 0.6  # Parámetro mu[span_4](start_span)[span_4](end_span)
sigma_cuadrada = 0.25  # Varianza sigma^2[span_5](start_span)[span_5](end_span)
sigma = np.sqrt(sigma_cuadrada)  # Desviación estándar (0.5)
t = 4  # Tiempo t[span_6](start_span)[span_6](end_span)

# Paso 1: N caminatas, registramos X_4
# Como X_t = mu*t + sigma*B_t, y B_t ~ Normal(0, t),
# entonces X_4 se distribuye Normal con media (mu * t) y varianza (sigma^2 * t).
N = 100000  # Número de caminatas (número de simulaciones)

media_teorica = mu * t  # 0.6 * 4 = 2.4
varianza_teorica = sigma_cuadrada * t  # 0.25 * 4 = 1.0
desviacion_teorica = np.sqrt(varianza_teorica)  # 1.0

# Generamos N valores aleatorios de X_4 usando una distribución normal
X_4 = np.random.normal(loc=media_teorica, scale=desviacion_teorica, size=N)

# Paso 2: Contamos caminatas con X_4 en el intervalo [1, 3]: M
a = 1
b = 3
condicion = (X_4 >= a) & (X_4 <= b)
M = np.sum(condicion)  # Cantidad de caminatas que cumplen la condición

# Paso 3: P = M / N
P = M / N

print(f"Número de caminatas (N): {N}")
print(f"Caminatas exitosas en [1, 3] (M): {M}")
print(f"Probabilidad estimada (P = M / N): {P:.5f}")
```
