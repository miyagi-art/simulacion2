```python
import random

def es_buena(secuencia):
    """Verifica que no existan '1's adyacentes."""
    for i in range(len(secuencia) - 1):
        if secuencia[i] == 1 and secuencia[i + 1] == 1:
            return False
    return True

def simular_cantidad_unos(m=100, num_pasos=100000, periodo_calentamiento=2000):
    # Comenzar con una secuencia válida de m ceros
    estado = [0] * m

    # Calentamiento para que la cadena alcance la distribución estacionaria
    for _ in range(periodo_calentamiento):
        idx = random.randint(0, m - 1)
        if estado[idx] == 1:
            estado[idx] = 0
        else:
            estado[idx] = 1
            if not es_buena(estado):
                estado[idx] = 0

    total_unos = 0
    min_unos = m
    max_unos = 0

    # Simulación principal
    for _ in range(num_pasos):
        idx = random.randint(0, m - 1)
        if estado[idx] == 1:
            estado[idx] = 0
        else:
            estado[idx] = 1
            if not es_buena(estado):
                estado[idx] = 0

        # Contar cuántos 1s tiene la cadena actual
        unos_actuales = sum(estado)
        total_unos += unos_actuales

        if unos_actuales < min_unos:
            min_unos = unos_actuales
        if unos_actuales > max_unos:
            max_unos = unos_actuales

    promedio_unos = total_unos / num_pasos
    return promedio_unos, min_unos, max_unos

# ==========================================
# EJECUCIÓN PARA m = 100
# ==========================================
m = 100
pasos = 100000

promedio, minimo, maximo = simular_cantidad_unos(m=m, num_pasos=pasos)

print(f"--- Simulación MCMC para m = {m} ---")
print(f"Promedio de 1's en las cadenas buenas (μ): {promedio:.2f}")
print(f"Mínimo de 1's observado: {minimo}")
print(f"Máximo de 1's observado: {maximo}")
```

```python
import random

def es_cadena_buena(secuencia):
    """
    Verifica si una secuencia es buena (sin unos adyacentes).
    """
    for i in range(len(secuencia) - 1):
        if secuencia[i] == 1 and secuencia[i + 1] == 1:
            return False
    return True

def simular_cadena_markov(m=4, pasos=100):
    # Estado inicial: secuencia de ceros de longitud m (siempre es una cadena buena)
    estado_actual = [0] * m

    # Historial para guardar el número de 1s en cada paso
    conteo_unos = [sum(estado_actual)]

    for _ in range(pasos):
        # 1. Elegir uno de sus m componentes de manera uniforme al azar
        idx = random.randint(0, m - 1)

        # Copiamos el estado actual para evaluar el cambio
        nuevo_estado = list(estado_actual)

        # 2. Si la componente es 1, cambiar a 0
        if estado_actual[idx] == 1:
            nuevo_estado[idx] = 0
            estado_actual = nuevo_estado # Siempre resulta en una cadena buena

        # 3. Si la componente es 0, cambiar a 1 solo si la secuencia resultante es buena
        else:
            nuevo_estado[idx] = 1
            if es_cadena_buena(nuevo_estado):
                estado_actual = nuevo_estado # Transición a la nueva secuencia
            # De lo contrario, se queda en el mismo estado (stay put)

        # Guardar la cantidad de unos en la secuencia actual
        conteo_unos.append(sum(estado_actual))


    return estado_actual, conteo_unos


# --- EJECUCIÓN DE LA SIMULACIÓN ---
m = 4 # Longitud de la secuencia
pasos = 100 # Número de iteraciones

estado_final, historial_unos = simular_cadena_markov(m=m, pasos=pasos)


print(f"--- Simulación a {pasos} pasos (m = {m}) ---")
print(f"Secuencia final alcanzada: {estado_final}")
print(f"Cantidad de unos en la secuencia final: {sum(estado_final)}")
print(f"Promedio de unos a lo largo de las {pasos} iteraciones: {sum(historial_unos) / len(historial_unos):.2f}")
```
