```python
import numpy as np
import matplotlib.pyplot as plt

def simular_movimiento_browniano(T, L):
    """
    Simula el movimiento browniano discretizado en el intervalo [0, T].
    """
    # 1. dt = T / L
    dt = T / L
    t = np.linspace(0, T, L + 1)

    # 2. W_0 = 0
    W = np.zeros(L + 1)
    W[0] = 0

    # 3. for i = 0 to L-1
    for i in range(L):
        # 4. calcular un xi_i ~ N(0,1)
        xi = np.random.normal(0, 1)

        # 5. hacer W_{i+1} = W_i + \delta W_i
        # donde \delta W_i = \sqrt{\delta t} * \xi_i
        W[i + 1] = W[i] + np.sqrt(dt) * xi

    # 6. Fin
    return t, W

# Parámetros de ejemplo
T = 1.0  # Tiempo final
L = 1000 # Número de pasos (L en los naturales)

t, W = simular_movimiento_browniano(T, L)

# Opcional: Gráfica para visualizar el resultado de la simulación
plt.figure(figsize=(10, 5))
plt.plot(t, W, label='Movimiento Browniano W(t)', color='blue')
plt.title('Simulación de Movimiento Browniano Discretizado')
plt.xlabel('t')
plt.ylabel('W(t)')
plt.grid(True)
plt.legend()
plt.show()
```
