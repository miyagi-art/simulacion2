```python
def generador (a, b, m, x0):
    x1 = (a*x0+b)%m
    return x1

a = 7
b = 5
m = 13
x0 = 3
n = 10
muestra = []
for i in range (n):
    nuevo = generador (a, b, m, x0)
    muestra.append(nuevo)
    x0 = nuevo
muestra
```
