¿cual es la probabilidad de que a largo plazo un dia sea soleado? acompañe su respuesta de una grafica.

```python
#probabilidad a largo plazo
p=Matrix([[1,0]])
for i in range (100):
  p=p*P
p
```

```python
import matplotlib.pyplot as plt
p=Matrix([[1,0]])
dias=[]
prob_soleado=[]
for i in range (50):
  dias.append(i)
  prob_soleado.append(float(p[0,0]))
  p=p*P

plt.plot(dias,prob_soleado)
plt.axhline(y=1/3, linestyle='--')
plt.xlabel("Dias")
plt.ylabel("Probabilidad de dia soleado")
plt.title("Probabilidad de un dia soleado a largo plazo")
plt.grid()
plt.show()
```
